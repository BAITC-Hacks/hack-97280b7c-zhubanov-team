# HackAlem AI — стартовый ноутбук

Ноутбук проверяет окружение и содержит безопасный шаблон подключения к OpenAI API. Секреты должны храниться только в переменных окружения или в локальном `.env`, который не коммитится.

```python
import sys
from importlib.metadata import version, PackageNotFoundError

print('Python:', sys.version)
print('Executable:', sys.executable)
```

```python
packages = ['numpy', 'pandas', 'scipy', 'sklearn', 'matplotlib', 'seaborn', 'plotly', 'requests', 'dotenv', 'pydantic', 'openai']
for package in packages:
    try:
        print(f'{package}: {version(package)}')
    except PackageNotFoundError:
        print(f'{package}: NOT INSTALLED')
```

```python
import numpy as np
import pandas as pd
import requests
from dotenv import load_dotenv

print('Base imports: OK')
print('NumPy test:', np.array([1, 2, 3]).mean())

optional = {}
for name, importer in {
    'scipy': lambda: __import__('scipy'),
    'sklearn': lambda: __import__('sklearn'),
    'matplotlib': lambda: __import__('matplotlib'),
    'seaborn': lambda: __import__('seaborn'),
    'plotly': lambda: __import__('plotly'),
}.items():
    try:
        importer(); optional[name] = 'available'
    except ImportError:
        optional[name] = 'missing — install from requirements.txt'
print('Optional scientific stack:', optional)
```

```python
import os

load_dotenv()
openai_key_present = bool(os.getenv('OPENAI_API_KEY'))
nvidia_key_present = bool(os.getenv('NVIDIA_API_KEY'))
print('OPENAI_API_KEY:', 'found' if openai_key_present else 'not set')
print('NVIDIA_API_KEY:', 'found' if nvidia_key_present else 'not set')
print('Keys are never printed.')
```

```python
# Безопасный шаблон: выполнится только если OPENAI_API_KEY задана в окружении.
if os.getenv('OPENAI_API_KEY'):
    from openai import OpenAI
    client = OpenAI()
    print('OpenAI client initialized.')
else:
    print('Set OPENAI_API_KEY before using OpenAI API.')
```
